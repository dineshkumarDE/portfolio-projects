```python
%sql
SELECT
    order_year,
    category,
    SUM(total_profit) AS category_profit
FROM
    presentation.ecomm_profit_agg
GROUP BY
    order_year,
    category
ORDER BY
    order_year, category;
```

```python
%sql
SELECT
    order_year,
    sub_category,
    SUM(total_profit) AS category_profit
FROM
    presentation.ecomm_profit_agg
GROUP BY
    order_year,
    sub_category
ORDER BY
    order_year, sub_category;
```

```python
%sql
SELECT
    order_year,
    SUM(total_profit) AS yearly_profit
FROM
    presentation.ecomm_profit_agg
GROUP BY
    order_year
ORDER BY
    order_year;
```
