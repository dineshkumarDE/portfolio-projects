### SQL Data Analysis Tasks



**1. Profit by year**


```python
%sql
select order_year,round(sum(total_profit), 2) as profit_per_year from presentation.ecomm_profit_agg group by order_year order by order_year
```

**2. Profit by year and category**


```python
%sql
select order_year,category,round(sum(total_profit), 2) as profit_per_year from presentation.ecomm_profit_agg group by order_year,category order by order_year,category
```

**3. profit by year and customerid**

```python
%sql
select order_year,customer_id,round(sum(total_profit), 2) as profit_per_year from presentation.ecomm_profit_agg group by order_year,customer_id order by order_year,customer_id
```

**4. profit by customer_id**

```python

%sql
select customer_id,round(sum(total_profit), 2) as profit from presentation.ecomm_profit_agg group by customer_id order by customer_id
```
